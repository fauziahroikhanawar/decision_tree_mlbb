```python
import pandas as pd
import numpy as np

# Set random seed untuk hasil yang konsisten
np.random.seed(42)

# Generate 30 samples of Mobile Legends match data
def generate_mlbb_data():
    data = []
    
    for i in range(30):
        # Basic statistics
        kda_ratio = round(np.random.uniform(0.5, 5.0), 2)
        gold_earned = np.random.randint(8000, 25000)
        turrets_destroyed = np.random.randint(0, 11)
        turret_damage = np.random.randint(2000, 15000)
        lord_kills = np.random.randint(0, 4)
        turtle_kills = np.random.randint(0, 6)
        
        # Team composition (ensure total = 5)
        # Generate random composition
        tank_count = np.random.randint(0, 3)
        fighter_count = np.random.randint(0, 3)
        assassin_count = np.random.randint(0, 3)
        mage_count = np.random.randint(0, 3)
        marksman_count = np.random.randint(0, 3)
        support_count = np.random.randint(0, 3)
        
        # Adjust to ensure total = 5
        current_total = tank_count + fighter_count + assassin_count + mage_count + marksman_count + support_count
        
        if current_total < 5:
            # Add to fighter if total < 5
            fighter_count += (5 - current_total)
        elif current_total > 5:
            # Reduce from the highest count
            counts = [tank_count, fighter_count, assassin_count, mage_count, marksman_count, support_count]
            while sum(counts) > 5:
                max_idx = np.argmax(counts)
                if counts[max_idx] > 0:
                    counts[max_idx] -= 1
            tank_count, fighter_count, assassin_count, mage_count, marksman_count, support_count = counts
        
        # Determine win based on multiple factors
        win_score = 0
        
        # KDA contribution
        if kda_ratio >= 2.5:
            win_score += 3
        elif kda_ratio >= 1.5:
            win_score += 2
        elif kda_ratio >= 1.0:
            win_score += 1
        
        # Gold contribution
        if gold_earned >= 20000:
            win_score += 3
        elif gold_earned >= 15000:
            win_score += 2
        elif gold_earned >= 12000:
            win_score += 1
            
        # Turrets contribution
        if turrets_destroyed >= 8:
            win_score += 3
        elif turrets_destroyed >= 6:
            win_score += 2
        elif turrets_destroyed >= 4:
            win_score += 1
            
        # Turret damage contribution
        if turret_damage >= 12000:
            win_score += 2
        elif turret_damage >= 8000:
            win_score += 1
            
        # Objectives contribution
        if lord_kills >= 2:
            win_score += 2
        elif lord_kills >= 1:
            win_score += 1
            
        if turtle_kills >= 4:
            win_score += 2
        elif turtle_kills >= 2:
            win_score += 1
        
        # Team composition bonus
        if tank_count >= 1 and marksman_count >= 1:  # Balanced team
            win_score += 1
        if mage_count >= 1:  # Magic damage
            win_score += 1
        
        # Determine win (with some randomness)
        base_win_prob = min(win_score / 15.0, 0.9)  # Max 90% win rate
        random_factor = np.random.random()
        kemenangan = 1 if random_factor < base_win_prob else 0
        
        # Create data row
        row = {
            'kda_ratio': kda_ratio,
            'gold_earned': gold_earned,
            'turrets_destroyed': turrets_destroyed,
            'turret_damage': turret_damage,
            'lord_kills': lord_kills,
            'turtle_kills': turtle_kills,
            'tank_count': tank_count,
            'fighter_count': fighter_count,
            'assassin_count': assassin_count,
            'mage_count': mage_count,
            'marksman_count': marksman_count,
            'support_count': support_count,
            'kemenangan': kemenangan
        }
        
        data.append(row)
    
    return pd.DataFrame(data)

# Generate the data
df = generate_mlbb_data()

# Display sample data
print("Sample MLBB Data (30 entries):")
print("="*50)
print(df.head(10))
print("\nData Info:")
print(f"Total samples: {len(df)}")
print(f"Win rate: {df['kemenangan'].mean():.2%}")
print(f"Features: {list(df.columns[:-1])}")

# Save to CSV
df.to_csv('mlbb_data_sample.csv', index=False)
print(f"\nData saved to 'mlbb_data_sample.csv'")

# Show some statistics
print("\nStatistics Summary:")
print("="*50)
print(df.describe())

# Show team composition distribution
print("\nTeam Composition Analysis:")
print("="*50)
for col in ['tank_count', 'fighter_count', 'assassin_count', 'mage_count', 'marksman_count', 'support_count']:
    print(f"{col}: {df[col].value_counts().sort_index().to_dict()}")

# Verify all teams have 5 heroes
team_totals = df[['tank_count', 'fighter_count', 'assassin_count', 'mage_count', 'marksman_count', 'support_count']].sum(axis=1)
print(f"\nTeam composition verification:")
print(f"All teams have 5 heroes: {all(team_totals == 5)}")
print(f"Team size distribution: {team_totals.value_counts()}")

print("\nCSV file created successfully! You can now use this with your MLBB predictor app.")
```
